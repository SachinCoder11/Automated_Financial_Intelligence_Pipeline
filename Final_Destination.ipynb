```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", 150)
pd.set_option("display.max_rows", 100)

FILE = r"XBRL-XBRL-PL-2025_part1.csv"

df = pd.read_csv(FILE, low_memory=False)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Total Cells:", df.size)

print("\nDATA TYPES")
print(df.dtypes.value_counts())

print("\nFIRST 5 ROWS")
display(df.head())
```

```python
missing = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing %": (df.isna().mean() * 100).round(2),
    "Data Type": df.dtypes
})

missing = missing.sort_values("Missing %", ascending=False)

display(missing)
```

```python
empty_cols = [col for col in df.columns if df[col].isna().all()]

print("Completely empty columns:", len(empty_cols))

for col in empty_cols:
    print(col)
```

```python
df_clean = df.drop(columns=empty_cols)

print("Columns before:", df.shape[1])
print("Columns after :", df_clean.shape[1])
```

```python
duplicates = df_clean.duplicated().sum()

print("Duplicate rows:", duplicates)
```

```python
print("CURRENT / PREVIOUS")
print(df_clean["current_previous"].value_counts(dropna=False))

print("\nSTART DATES")
display(df_clean["DateStartReportingPeriod"].value_counts().head(20))

print("\nEND DATES")
display(df_clean["DateEndReportingPeriod"].value_counts().head(20))
```

```python
fy = df_clean[
    (df_clean["current_previous"].astype(str).str.strip() == "current") &
    (df_clean["DateStartReportingPeriod"] == "2024-04-01") &
    (df_clean["DateEndReportingPeriod"] == "2025-03-31")
].copy()

print("Raw rows:", len(df_clean))
print("FY 2024-25 rows:", len(fy))
```

```python
important_columns = [
    "company_uid",
    "Revenue",
    "RevenueFromOperations",
    "Expenses",
    "ProfitBeforeTax",
    "TaxExpense",
    "ProfitLossForPeriod",
    "FinanceCosts",
    "EmployeeBenefitExpense",
    "DepreciationExpense",
    "OtherExpenses",
    "ResearchDevelopmentExpenditure",
    "CSRExpenditure"
]

available = [col for col in important_columns if col in fy.columns]
missing_important = [col for col in important_columns if col not in fy.columns]

print("AVAILABLE:")
print(available)

print("\nNOT FOUND:")
print(missing_important)
```

```python
financial_cols = [
    "Revenue",
    "RevenueFromOperations",
    "Expenses",
    "ProfitBeforeTax",
    "TaxExpense",
    "ProfitLossForPeriod",
    "FinanceCosts",
    "EmployeeBenefitExpense",
    "DepreciationExpense",
    "OtherExpenses",
    "ResearchDevelopmentExpenditure",
    "CSRExpenditure"
]

financial_cols = [col for col in financial_cols if col in fy.columns]

for col in financial_cols:
    fy[col] = pd.to_numeric(fy[col], errors="coerce")

print(fy[financial_cols].dtypes)
```

```python
financial_summary = fy[financial_cols].describe().T

financial_summary["Missing %"] = (
    fy[financial_cols].isna().mean() * 100
).round(2)

display(financial_summary)
```

```python
print("Revenue > 0 :", (fy["Revenue"] > 0).sum())
print("Revenue = 0 :", (fy["Revenue"] == 0).sum())
print("Revenue < 0 :", (fy["Revenue"] < 0).sum())
print("Revenue NULL:", fy["Revenue"].isna().sum())
```

```python
print("Profitable:", (fy["ProfitLossForPeriod"] > 0).sum())
print("Loss-making:", (fy["ProfitLossForPeriod"] < 0).sum())
print("Zero profit:", (fy["ProfitLossForPeriod"] == 0).sum())
print("Missing:", fy["ProfitLossForPeriod"].isna().sum())
```

```python
analysis = fy[fy["Revenue"] > 0].copy()

analysis["Profit_Margin"] = (
    analysis["ProfitLossForPeriod"]
    / analysis["Revenue"]
) * 100

analysis["Expense_Ratio"] = (
    analysis["Expenses"]
    / analysis["Revenue"]
) * 100

analysis["Finance_Cost_Ratio"] = (
    analysis["FinanceCosts"]
    / analysis["Revenue"]
) * 100

analysis["Employee_Cost_Ratio"] = (
    analysis["EmployeeBenefitExpense"]
    / analysis["Revenue"]
) * 100

analysis["R&D_Intensity"] = (
    analysis["ResearchDevelopmentExpenditure"]
    / analysis["Revenue"]
) * 100
```

```python
margin_stats = analysis["Profit_Margin"].describe()

print(margin_stats)
```

```python
print(
    "Median Profit Margin:",
    analysis["Profit_Margin"].median()
)
```

```python
q1 = analysis["Profit_Margin"].quantile(0.25)
q3 = analysis["Profit_Margin"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = analysis[
    (analysis["Profit_Margin"] < lower) |
    (analysis["Profit_Margin"] > upper)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower bound:", lower)
print("Upper bound:", upper)

print("\nPotential outliers:", len(outliers))
```

```python
margin = analysis["Profit_Margin"].replace(
    [np.inf, -np.inf],
    np.nan
).dropna()

# Clip ONLY for visualization
low = margin.quantile(0.01)
high = margin.quantile(0.99)

margin_plot = margin.clip(low, high)

plt.figure(figsize=(10, 5))

sns.histplot(
    margin_plot,
    bins=50
)

plt.axvline(
    margin.median(),
    linewidth=2,
    label=f"Median = {margin.median():.2f}%"
)

plt.title("Profit Margin Distribution - FY 2024-25")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Number of Companies")
plt.legend()

plt.show()
```

```python
plot_data = analysis[
    analysis["Profit_Margin"].notna()
].copy()

# Remove extreme values ONLY for visualization
plot_data["Margin_Plot"] = plot_data["Profit_Margin"].clip(-100, 100)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=plot_data,
    x="Revenue",
    y="Margin_Plot",
    alpha=0.25,
    s=20
)

plt.xscale("log")

plt.axhline(0, linewidth=1)

plt.title("Revenue vs Profit Margin")
plt.xlabel("Revenue (Log Scale)")
plt.ylabel("Profit Margin (%)")

plt.show()
```

```python
finance_data = analysis[
    analysis["Finance_Cost_Ratio"].notna() &
    np.isfinite(analysis["Finance_Cost_Ratio"])
].copy()

finance_data["Finance_Cost_Burden"] = pd.qcut(
    finance_data["Finance_Cost_Ratio"].rank(method="first"),
    4,
    labels=[
        "Q1 - Low",
        "Q2",
        "Q3",
        "Q4 - High"
    ]
)
```

```python
grouped = (
    finance_data
    .groupby(
        "Finance_Cost_Burden",
        observed=False
    )["Profit_Margin"]
    .median()
)

plt.figure(figsize=(9, 5))

plt.bar(
    grouped.index.astype(str),
    grouped.values
)

plt.axhline(0, linewidth=1)

plt.title("Median Profit Margin by Finance Cost Burden")
plt.xlabel("Finance Cost / Revenue Quartile")
plt.ylabel("Median Profit Margin (%)")

plt.show()
```

```python
feature_cols = [
    "Revenue",
    "Expenses",
    "ProfitBeforeTax",
    "FinanceCosts",
    "EmployeeBenefitExpense",
    "DepreciationExpense",
    "OtherExpenses",
    "ProfitLossForPeriod",
    "Profit_Margin",
    "Expense_Ratio",
    "Finance_Cost_Ratio",
    "Employee_Cost_Ratio",
    "R&D_Intensity"
]

feature_cols = [
    col for col in feature_cols
    if col in analysis.columns
]

corr = analysis[feature_cols].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    center=0
)

plt.title("Financial Variable Correlation Matrix")

plt.show()
```

```python
numeric_cols = analysis.select_dtypes(
    include="number"
).columns

skewness = (
    analysis[numeric_cols]
    .skew()
    .sort_values(ascending=False)
)

display(
    skewness.to_frame("Skewness")
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Revenue data
revenue = analysis["Revenue"].dropna()

# Skewness calculate
skew_value = revenue.skew()

print("Revenue Skewness:", round(skew_value, 2))

if skew_value > 0:
    print("Distribution: RIGHT-SKEWED (Positive Skewness)")
elif skew_value < 0:
    print("Distribution: LEFT-SKEWED (Negative Skewness)")
else:
    print("Distribution: Approximately Symmetric")

# Plot
plt.figure(figsize=(10, 5))

sns.histplot(
    revenue,
    bins=100,
    kde=True
)

plt.title(
    f"Revenue Distribution | Skewness = {skew_value:.2f}"
)

plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

```python
financial_cols = [
    "Revenue",
    "Expenses",
    "ProfitBeforeTax",
    "FinanceCosts",
    "EmployeeBenefitExpense",
    "DepreciationExpense",
    "OtherExpenses",
    "ProfitLossForPeriod"
]

for col in financial_cols:

    if col not in analysis.columns:
        continue

    data = analysis[col].dropna()

    skew_value = data.skew()

    if skew_value > 0:
        direction = "RIGHT-SKEWED"
    elif skew_value < 0:
        direction = "LEFT-SKEWED"
    else:
        direction = "SYMMETRIC"

    print(
        f"{col:30} | "
        f"Skewness: {skew_value:8.2f} | "
        f"{direction}"
    )
```

```python
skew_results = []

for col in financial_cols:

    if col not in analysis.columns:
        continue

    value = analysis[col].dropna().skew()

    if value > 1:
        category = "Highly Right-Skewed"
    elif value > 0.5:
        category = "Moderately Right-Skewed"
    elif value < -1:
        category = "Highly Left-Skewed"
    elif value < -0.5:
        category = "Moderately Left-Skewed"
    else:
        category = "Approximately Symmetric"

    skew_results.append({
        "Column": col,
        "Skewness": round(value, 3),
        "Distribution": category
    })

skew_table = pd.DataFrame(skew_results)

display(skew_table)
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    analysis["ProfitLossForPeriod"].dropna(),
    bins=100
)

plt.title("Profit/Loss Distribution")
plt.xlabel("Profit / Loss")
plt.ylabel("Number of Companies")

plt.show()
```

```python
revenue = analysis.loc[
    analysis["Revenue"] > 0,
    "Revenue"
]

plt.figure(figsize=(10, 5))

sns.histplot(
    np.log10(revenue),
    bins=50
)

plt.title("Log-Transformed Revenue Distribution")
plt.xlabel("log10(Revenue)")
plt.ylabel("Number of Companies")

plt.show()
```

The financial variables exhibit substantial skewness. Revenue, expenses and major cost components are strongly right-skewed, indicating a highly uneven distribution of financial scale across observations. Profit/Loss is strongly left-skewed, indicating the presence of observations with exceptionally large losses. Therefore, percentile-based analysis and appropriate transformations are required before interpreting relationships or building predictive models.
